# Question 1 – Student Performance Analysis
---

## Part a) Import Libraries, Create DataFrame & Explain Structure

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    'Student_ID': range(1, 13),
    'Name': ['Alice','Bob','Carol','David','Eva','Frank',
             'Grace','Henry','Iris','Jack','Karen','Leo'],
    'Gender': ['F','M','F','M','F','M','F','M','F','M','F','M'],
    'Programming':    [85, 72, 90, 55, 68, 78, 92, 61, 74, 88, 45, 79],
    'Networking':     [78, 65, 82, 70, 55, 88, 75, 60, 85, 72, 58, 83],
    'Database':       [90, 80, 75, 65, 72, 68, 88, 70, 79, 65, 55, 74],
    'Cyber_Security': [70, 68, 85, 60, 78, 72, 80, 55, 90, 75, 62, 68]
}

df = pd.DataFrame(data)
print("First 10 rows:")
df.head(10)
```

```python
print("Shape:", df.shape)
print("\nData Types:")
print(df.dtypes)
print("\nBasic Info:")
df.info()
```

### Dataset Structure Explanation

The dataset has **12 rows** (one per student) and **7 columns**:

| Column | Type | Description |
|---|---|---|
| `Student_ID` | int64 | Unique integer identifier for each student |
| `Name` | object | Student's full name (string) |
| `Gender` | object | Categorical – 'M' (Male) or 'F' (Female) |
| `Programming` | int64 | Score out of 100 in Programming |
| `Networking` | int64 | Score out of 100 in Networking |
| `Database` | int64 | Score out of 100 in Database |
| `Cyber_Security` | int64 | Score out of 100 in Cyber Security |

- `df.shape` returns **(12, 7)** confirming 12 students and 7 features.
- Numeric subject columns have dtype `int64`; Name and Gender have dtype `object`.
- There are no missing values in this clean synthetic dataset.

## Part b) Average Score, Grade Assignment & Grade Count

```python
# Compute average score per student
df['Average'] = df[['Programming', 'Networking', 'Database', 'Cyber_Security']].mean(axis=1)

# Grade assignment function
def assign_grade(avg):
    if avg >= 80:   return 'A'
    elif avg >= 70: return 'B'
    elif avg >= 60: return 'C'
    elif avg >= 50: return 'D'
    else:           return 'F'

df['Grade'] = df['Average'].apply(assign_grade)

print("Student Averages and Grades:")
print(df[['Name', 'Average', 'Grade']].to_string(index=False))

print("\nGrade Distribution:")
grade_counts = df['Grade'].value_counts().sort_index()
print(grade_counts)
```

## Part c) Visualizations & Interpretations

```python
# Bar Chart – Grade Distribution
colors = {'A': '#2ecc71', 'B': '#3498db', 'C': '#f39c12', 'D': '#e67e22', 'F': '#e74c3c'}
bar_colors = [colors.get(g, '#999') for g in grade_counts.index]

plt.figure(figsize=(7, 4))
grade_counts.plot(kind='bar', color=bar_colors, edgecolor='white', width=0.6)
plt.title('Grade Distribution of Students', fontsize=14, fontweight='bold')
plt.xlabel('Grade', fontsize=12)
plt.ylabel('Number of Students', fontsize=12)
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

### Bar Chart Interpretation
The bar chart shows the frequency of each letter grade in the class. If most students fall in the B–C range, the class performance is moderate. A tall 'A' bar indicates high achievement across subjects, while a prominent 'F' bar signals students requiring urgent academic support. This visual instantly communicates the overall health of classroom performance to educators and administrators.

```python
# Boxplot – Average Score by Gender
plt.figure(figsize=(7, 4))
sns.boxplot(x='Gender', y='Average', data=df, palette='Set2', width=0.5)
plt.title('Average Score Distribution by Gender', fontsize=14, fontweight='bold')
plt.xlabel('Gender', fontsize=12)
plt.ylabel('Average Score', fontsize=12)
plt.tight_layout()
plt.show()
```

### Boxplot Interpretation
The boxplot shows the **median** (line inside box), **interquartile range** (box height = IQR = Q3–Q1), and **outliers** (dots beyond whiskers) for each gender group.
- A higher median line for one gender suggests that group performs better on average.
- A wider box means greater variability in scores within that group.
- Outlier dots represent exceptional over- or under-performers.

This comparison helps identify gender-based performance gaps and guides targeted interventions.

# Question 2 – Agricultural Dataset Cleaning
---

## Part a) Generating Synthetic Data with NumPy

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)  # Ensures reproducibility
n = 100

df_agri = pd.DataFrame({
    'Plot_ID': range(1, n + 1),
    'Region':      np.random.choice(['North', 'South', 'East', 'West'], n),
    'Maize_Yield': np.random.normal(4.5, 2.5, n),   # tons/hectare – includes negatives & outliers
    'Rice_Yield':  np.random.normal(3.8, 2.2, n)
})

print("Generated Dataset (first 10 rows):")
print(df_agri.head(10).to_string(index=False))
print(f"\nNegative Maize yields: {(df_agri['Maize_Yield'] < 0).sum()}")
print(f"Maize yields > 10 t/ha: {(df_agri['Maize_Yield'] > 10).sum()}")
```

### Explanation – NumPy Random Functions Used

| Function | Description |
|---|---|
| `np.random.seed(42)` | Seeds the RNG for reproducibility across runs |
| `np.random.normal(mean, std, size)` | Generates values from a Gaussian distribution – models yield data that clusters around a typical value |
| `np.random.choice(array, size)` | Randomly selects categorical values – used to assign regions |
| `np.random.uniform(low, high, size)` | Alternative – generates uniformly distributed values within a range |

Synthetic data mimics real-world data without privacy concerns, making it ideal for testing data pipelines.

## Part b) Data Quality Problems & Their Impact on National Planning

### Problem 1 – Negative Yield Values
A negative crop yield is **physically impossible**. These arise from random number generation or data entry errors. If included in national food security estimates, they artificially reduce totals, causing governments to:
- Underestimate food production
- Misallocate agricultural subsidies and aid
- Make incorrect import/export decisions

### Problem 2 – Extreme Outliers (Yields > 10 tons/hectare)
Under typical African smallholder farming conditions, staple crop yields rarely exceed 10 t/ha. Outliers caused by data entry errors or sensor faults can **inflate mean values**, leading planners to:
- Believe food security is better than reality
- Reduce investment in farming infrastructure
- Fail to prepare for food shortages — with serious humanitarian consequences

## Part c) Data Cleaning Pipeline & Visualization

```python
print("Before cleaning:")
print(df_agri[['Maize_Yield', 'Rice_Yield']].describe().round(2))

# Step 1: Replace negative values with NaN
df_agri['Maize_Yield'] = df_agri['Maize_Yield'].where(df_agri['Maize_Yield'] >= 0, np.nan)
df_agri['Rice_Yield']  = df_agri['Rice_Yield'].where(df_agri['Rice_Yield']  >= 0, np.nan)

# Step 2: Remove yields above 10 tons/hectare (replace with NaN)
df_agri.loc[df_agri['Maize_Yield'] > 10, 'Maize_Yield'] = np.nan
df_agri.loc[df_agri['Rice_Yield']  > 10, 'Rice_Yield']  = np.nan

print(f"\nMissing after Steps 1 & 2 – Maize: {df_agri['Maize_Yield'].isna().sum()}, Rice: {df_agri['Rice_Yield'].isna().sum()}")

# Step 3: Impute missing values using column mean
df_agri['Maize_Yield'].fillna(df_agri['Maize_Yield'].mean(), inplace=True)
df_agri['Rice_Yield'].fillna(df_agri['Rice_Yield'].mean(),   inplace=True)

print("\nAfter cleaning:")
print(df_agri[['Maize_Yield', 'Rice_Yield']].describe().round(2))
```

```python
# Step 4: Plot number of farms per region
region_counts = df_agri['Region'].value_counts()

plt.figure(figsize=(7, 4))
region_counts.plot(kind='bar', color='steelblue', edgecolor='white', width=0.6)
plt.title('Number of Farms per Region', fontsize=14, fontweight='bold')
plt.xlabel('Region', fontsize=12)
plt.ylabel('Count', fontsize=12)
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

### Interpretation
After cleaning, the dataset contains only realistic yield values (0–10 t/ha). **Mean imputation** preserves dataset size while eliminating NaN values that would disrupt calculations. The bar chart of farms per region reveals whether data collection was geographically balanced. Unequal representation (e.g., far more Southern farms) means national averages are skewed towards that region's conditions, potentially misleading policy decisions for underrepresented areas.

# Question 3 – Fitness Tracker EDA
---

```python
# Create fitness tracker dataset
np.random.seed(10)
n = 200

df_fit = pd.DataFrame({
    'daily_steps':    np.random.normal(8000, 2500, n).astype(int),
    'calories':       np.random.normal(2200, 400, n),
    'heart_rate':     np.random.normal(72, 12, n),
    'sleep_hours':    np.random.normal(7, 1.2, n),
    'activity_level': np.random.choice(['Low', 'Moderate', 'High'], n)
})

print("Dataset Preview:")
df_fit.head()
```

## Part a) Mean and Standard Deviation of Daily Steps

```python
mean_steps = df_fit['daily_steps'].mean()
std_steps  = df_fit['daily_steps'].std()
cv         = (std_steps / mean_steps) * 100  # Coefficient of variation

print(f"Mean daily steps : {mean_steps:,.1f}")
print(f"Std  daily steps : {std_steps:,.1f}")
print(f"Coefficient of variation: {cv:.1f}%")
```

### Interpretation of Variability
- The **mean** tells us the typical number of daily steps across all users.
- The **standard deviation** measures how spread out individual step counts are from that average.
- A high standard deviation (≈ 2500 steps) indicates **substantial variability** – some users are very sedentary while others are highly active.
- The **coefficient of variation** (std/mean × 100) contextualises this spread as a percentage; a CV above 30% signals high relative variability.
- About **68%** of users fall within one standard deviation of the mean (mean ± std), and **95%** within two standard deviations.

## Part b) Correlation Between Daily Steps and Calories

```python
from scipy import stats

correlation = df_fit['daily_steps'].corr(df_fit['calories'])
print(f"Pearson Correlation (steps vs calories): {correlation:.3f}")

# Scatter plot with regression line
plt.figure(figsize=(8, 5))
sns.regplot(
    x='daily_steps', y='calories', data=df_fit,
    scatter_kws={'alpha': 0.4, 'color': 'steelblue', 's': 30},
    line_kws={'color': 'red', 'linewidth': 2}
)
plt.title('Daily Steps vs Calories Burned', fontsize=14, fontweight='bold')
plt.xlabel('Daily Steps', fontsize=12)
plt.ylabel('Calories Burned', fontsize=12)
plt.tight_layout()
plt.show()
```

### Interpretation – Strength and Direction
- **Direction**: A positive correlation means that as daily steps increase, calories burned also increase — physiologically expected since walking and running are major contributors to energy expenditure.
- **Strength**: 
  - |r| close to 1.0 → strong relationship
  - |r| around 0.5 → moderate relationship
  - |r| close to 0 → weak/no linear relationship
- The **red regression line** visually confirms the upward trend. Points scattered closely around the line indicate a strong relationship; wide scatter indicates a weaker one.
- Note: Correlation does not imply causation — other factors (metabolism, exercise type) also influence calories burned.

## Part c) Average Sleep Hours by Activity Level

```python
sleep_by_activity = df_fit.groupby('activity_level')['sleep_hours'].mean().round(2)
print("Average Sleep Hours by Activity Level:")
print(sleep_by_activity)

# Bar chart for clarity
sleep_by_activity.plot(kind='bar', color=['#e74c3c','#f39c12','#2ecc71'], edgecolor='white', figsize=(6,4))
plt.title('Average Sleep Hours by Activity Level', fontsize=13, fontweight='bold')
plt.xlabel('Activity Level')
plt.ylabel('Avg Sleep Hours')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

### Findings
Grouping by `activity_level` and computing mean sleep hours reveals important health patterns:
- **High activity** individuals may sleep more due to physical recovery needs, or less if they are time-poor.
- **Low activity** users may show irregular or reduced sleep, often linked to sedentary lifestyle habits.
- **Moderate activity** is frequently associated with optimal sleep quality in health research.

These insights help fitness app developers send personalised nudges — for example, prompting low-activity users to improve both movement and sleep habits simultaneously.

# Question 4 – Real Estate Market Analysis
---

## Part a) Load Dataset & Summary Statistics

```python
np.random.seed(5)
n = 150

df_re = pd.DataFrame({
    'price':      np.random.normal(350000, 120000, n),
    'location':   np.random.choice(['Urban', 'Suburban', 'Rural'], n),
    'size_sqft':  np.random.normal(1800, 500, n),
    'bedrooms':   np.random.choice([1, 2, 3, 4, 5], n),
    'bathrooms':  np.random.choice([1, 2, 3], n)
})

print("Summary Statistics:")
df_re.describe().round(2)
```

### Identifying Pricing Variables
- **`price`** – Primary outcome variable; the target for any predictive model.
- **`size_sqft`** – Strong continuous predictor; larger homes generally cost more.
- **`bedrooms` / `bathrooms`** – Count variables that influence value; more rooms typically increase price.
- **`location`** – Categorical predictor; Urban properties usually command a premium over Rural ones.

The `describe()` output reveals the range, spread (std), and potential outliers (compare min/max to mean ± 2×std) for each variable.

## Part b) Handle Missing Values & Create price_per_sqft Feature

```python
# Introduce missing values to simulate real-world data
missing_idx_price = np.random.choice(df_re.index, 10, replace=False)
missing_idx_size  = np.random.choice(df_re.index, 8, replace=False)
df_re.loc[missing_idx_price, 'price']     = np.nan
df_re.loc[missing_idx_size,  'size_sqft'] = np.nan

print(f"Missing values – price: {df_re['price'].isna().sum()}, size_sqft: {df_re['size_sqft'].isna().sum()}")

# Impute with median (robust to outliers in financial data)
df_re['price'].fillna(df_re['price'].median(),         inplace=True)
df_re['size_sqft'].fillna(df_re['size_sqft'].median(), inplace=True)

print(f"After imputation – missing: {df_re.isna().sum().sum()}")

# Create derived feature
df_re['price_per_sqft'] = df_re['price'] / df_re['size_sqft']

print(f"\nAverage price per sqft: ${df_re['price_per_sqft'].mean():.2f}")
df_re[['price', 'size_sqft', 'price_per_sqft']].head()
```

**Why median imputation?** Median is preferred over mean for financial data because extreme high-value properties (luxury homes) skew the mean upward. The median is resistant to such outliers, producing a more representative typical value for imputation.

**`price_per_sqft`** normalises property value for fair comparison across different sizes — it is a standard metric in real estate valuation used to compare otherwise incomparable properties.

## Part c) Visualizations & Interpretations

```python
# Histogram of Prices
plt.figure(figsize=(8, 4))
df_re['price'].hist(bins=25, color='#3498db', edgecolor='white')
plt.axvline(df_re['price'].mean(),   color='red',    linestyle='--', label=f'Mean: ${df_re["price"].mean():,.0f}')
plt.axvline(df_re['price'].median(), color='orange', linestyle='--', label=f'Median: ${df_re["price"].median():,.0f}')
plt.title('Distribution of Property Prices', fontsize=14, fontweight='bold')
plt.xlabel('Price ($)')
plt.ylabel('Frequency')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Scatter Plot – Size vs Price
plt.figure(figsize=(8, 5))
plt.scatter(df_re['size_sqft'], df_re['price'], alpha=0.5, color='#e74c3c', s=30)
m, b = np.polyfit(df_re['size_sqft'], df_re['price'], 1)
x_line = np.linspace(df_re['size_sqft'].min(), df_re['size_sqft'].max(), 100)
plt.plot(x_line, m * x_line + b, color='navy', linewidth=2, label=f'Trend line (slope: ${m:,.0f}/sqft)')
plt.title('Property Size vs Price', fontsize=14, fontweight='bold')
plt.xlabel('Size (sqft)')
plt.ylabel('Price ($)')
plt.legend()
plt.tight_layout()
plt.show()
```

### Interpretation of Patterns

**Histogram:** Real estate price distributions are typically **right-skewed** — most properties cluster in an affordable range while a long right tail represents luxury properties. When mean > median (mean is pulled right by high-value outliers), this confirms the skew. A bimodal shape could indicate two distinct submarkets (affordable vs. premium).

**Scatter Plot:** Shows a **positive relationship** — larger properties cost more. The trend line slope indicates price per additional square foot. Clusters of points at certain sizes (e.g., 1500–2000 sqft) reveal popular property sizes in the market. Outliers — expensive small properties — may indicate premium urban locations; cheap large properties may signal rural areas with low land value.

# Question 5 – Customer Feedback Data Cleaning
---

## Part a) Four Data Quality Issues in Multi-Source Datasets

### 1. Missing Values
When data is collected from multiple sources (web forms, call centres, emails), some fields are left blank. Missing data reduces the effective sample size and can bias results if the missing values are not random (MNAR – Missing Not At Random). For example, customers with very negative experiences may deliberately skip satisfaction ratings.

### 2. Duplicate Records
The same customer feedback may be submitted multiple times due to double-clicks, system retries, or overlapping database merges. Duplicates inflate counts and skew aggregate statistics like average satisfaction scores. They can also cause ML models to overfit to repeated records.

### 3. Inconsistent Formatting & Encoding
Different sources represent the same value differently:
- Dates: `'01/15/2024'` vs `'2024-01-15'`
- Gender: `'M'`/`'F'` vs `'Male'`/`'Female'`
- Ratings: 1–5 scale vs 0–10 scale

Without standardisation, `groupby()` and `merge()` operations produce incorrect or incomplete results.

### 4. Outliers and Invalid Entries
Human data entry errors produce impossible values like `age = 999`, `satisfaction = -3`, or dates set in the future. Automated systems may log sensor glitches. These outliers distort statistical summaries and mislead visualisations, making averages appear unrealistically high or low.

## Part b) Three Pandas Cleaning Techniques

```python
# Demonstration dataset with all four issues
df_feedback = pd.DataFrame({
    'customer_id': [1, 2, 2, 3, 4, 5, 5],
    'rating':      [4, None, None, 5, 999, 3, 3],
    'age':         [25, 30, 30, 22, 45, None, None],
    'gender':      ['Male', 'female', 'female', 'M', 'F', 'MALE', 'MALE'],
    'date':        ['2024-01-10', '2024-01-11', '2024-01-11', '2024-01-12', '2024-01-13', '2024-01-14', '2024-01-14']
})

print("Raw data:")
print(df_feedback.to_string(index=False))
```

```python
# 1. dropna() – Remove rows with missing values
df_no_na = df_feedback.dropna(subset=['rating'])  # Only drop where 'rating' is NaN
print("After dropna(subset=['rating']):")
print(df_no_na.to_string(index=False))
```

```python
# 2. fillna() – Impute missing values
df_filled = df_feedback.copy()
df_filled['rating'] = df_filled['rating'].fillna(df_filled['rating'].median())
df_filled['age']    = df_filled['age'].fillna(df_filled['age'].mean().round())
print("After fillna() – no more NaN:")
print(df_filled[['customer_id','rating','age']].to_string(index=False))
```

```python
# 3. drop_duplicates() – Remove duplicate rows
df_clean = df_filled.drop_duplicates(subset=['customer_id'], keep='first')
print("After drop_duplicates (by customer_id):")
print(df_clean.to_string(index=False))
```

```python
# 4. astype() – Fix data types and standardise gender
df_clean = df_clean.copy()
df_clean['rating'] = df_clean['rating'].astype(int)
df_clean['date']   = pd.to_datetime(df_clean['date'])
df_clean['gender'] = df_clean['gender'].str.lower().str.strip().replace({'m': 'male', 'f': 'female'})
print("After astype() and standardisation:")
print(df_clean.dtypes)
print(df_clean.to_string(index=False))
```

## Part c) How Poor Data Affects Analysis

### Effect on Mean & Variance
Outliers and erroneous values directly distort the **mean**, pulling it toward extreme values. A single entry of `age=999` in a 100-row dataset can shift the mean by nearly 10 years. **Variance** is even more sensitive — being based on squared deviations, one large outlier can inflate variance by orders of magnitude, making data appear far more variable than it truly is.

### Effect on ML Model Accuracy
- **Missing values** cause many algorithms (linear regression, SVMs) to fail without prior imputation.
- **Duplicates** cause overfitting — the model memorises repeated records rather than learning general patterns.
- **Outliers** dominate loss functions like MSE, causing models to learn inaccurate decision boundaries that optimise for rare extremes.
- **Inconsistent categories** (e.g., `'male'` vs `'Male'`) create spurious extra features after one-hot encoding, degrading model generalisation.

### Effect on Visualization Clarity
- **Outliers** compress the visible range of charts — a single value of 10,000 in a dataset where most values are 50–100 makes the histogram nearly unreadable.
- **Missing values** create gaps in time series plots.
- **Duplicate data** inflates bar heights in count charts, misrepresenting frequencies.
- Clean data ensures visualisations accurately communicate true patterns to decision-makers.

# Question 6 – Agricultural Yield Visualization
---

## Part a) Roles of NumPy, Pandas & Matplotlib in Agricultural Analytics

### NumPy – Numerical Foundation
NumPy provides fast array operations essential for large agricultural datasets. Functions like `np.mean()`, `np.std()`, and `np.percentile()` compute summary statistics efficiently across thousands of farm records. `np.random` functions generate synthetic test data. NumPy's array broadcasting enables element-wise transformations (e.g., converting kg/hectare → tons/hectare) without slow Python loops. It underpins both Pandas and Matplotlib internally.

### Pandas – Data Management
Pandas provides the **DataFrame** structure that mirrors a database table. In agricultural analytics it is used to:
- Load CSV files from national crop surveys
- Clean invalid yield entries with `fillna()` and `dropna()`
- Merge regional datasets with `merge()` and `concat()`
- Group data by season or region with `groupby()`
- Compute derived metrics like `yield_per_rainfall`

### Matplotlib – Visualisation Engine
Matplotlib translates data into visual charts for intuitive interpretation. In agricultural analytics it produces:
- **Histograms** of yield distributions
- **Bar charts** comparing regional production
- **Line plots** showing seasonal yield trends
- **Scatter plots** exploring rainfall–yield relationships

Its integration with Pandas (`df.plot()`) enables quick visualisation, and its customisation options allow publication-quality figures for government reports.

## Part b) Generate 60 Rainfall Values, Compute Statistics & Plot Histogram

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(7)
rainfall = np.random.uniform(500, 1500, 60)  # 60 random rainfall values (mm)

# Compute statistics
mean_rain   = np.mean(rainfall)
median_rain = np.median(rainfall)
std_rain    = np.std(rainfall)
min_rain    = np.min(rainfall)
max_rain    = np.max(rainfall)

print("Rainfall Statistics")
print("-" * 30)
print(f"Mean   : {mean_rain:.2f} mm")
print(f"Median : {median_rain:.2f} mm")
print(f"Std Dev: {std_rain:.2f} mm")
print(f"Min    : {min_rain:.2f} mm")
print(f"Max    : {max_rain:.2f} mm")
```

```python
# Histogram titled "Rainfall Distribution"
plt.figure(figsize=(8, 5))
plt.hist(rainfall, bins=10, color='#27ae60', edgecolor='white', alpha=0.85)
plt.axvline(mean_rain,   color='red',    linestyle='--', linewidth=2, label=f'Mean: {mean_rain:.1f} mm')
plt.axvline(median_rain, color='orange', linestyle=':',  linewidth=2, label=f'Median: {median_rain:.1f} mm')
plt.title('Rainfall Distribution', fontsize=15, fontweight='bold')
plt.xlabel('Rainfall (mm)', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.legend(fontsize=11)
plt.tight_layout()
plt.show()
```

### Interpretation
With `np.random.uniform(500, 1500, 60)`, values are approximately evenly spread across the range. The **expected mean ≈ 1000 mm** and the **theoretical standard deviation ≈ 289 mm** (formula: (max − min) / √12). The histogram displays a roughly **flat (uniform) distribution** with slight random variation. The red mean line and orange median line lie close together (as expected for a symmetric distribution). The vertical spread of bars across all bins confirms no particular rainfall amount dominates — all values between 500 and 1500 mm are equally likely.

## Part c) Two Advantages of Visualization Over Raw Numeric Tables

### Advantage 1 – Instant Pattern Recognition
The human visual system processes spatial patterns far faster than sequences of numbers. A histogram of 60 rainfall values communicates the **shape**, **centre**, and **spread** of the distribution in under a second. Scanning a table of 60 numbers to identify patterns would take minutes and is highly error-prone. Visualisations reveal skewness, bimodality, outliers, and trends at a glance — insights that are nearly invisible in raw tables. For agricultural planners reviewing hundreds of variables across multiple seasons, this speed advantage is invaluable.

### Advantage 2 – Effective Communication to Non-Technical Stakeholders
Policymakers, farmers, and investors may lack the statistical training to interpret tables of means, standard deviations, and percentiles. A bar chart showing that the Northern Region produces **2× more maize** than the Southern Region communicates this finding clearly to any audience. Visualisations transcend data literacy barriers, enabling evidence-based decision-making at all levels. Charts can be embedded in government reports, funding proposals, and extension service dashboards to drive real-world agricultural policy action.
