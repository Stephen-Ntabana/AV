## Question 2: Differentiate between Descriptive, Diagnostic, Predictive, and Prescriptive Analytics

| Type | Question Answered | Focus | Example |
|------|-------------------|-------|---------|
| **Descriptive** | *What happened?* | Summarizes historical data | Monthly sales report showing revenue trends |
| **Diagnostic** | *Why did it happen?* | Root cause analysis | Investigating why sales dropped in Q3 |
| **Predictive** | *What will happen?* | Forecasting future outcomes using ML/statistics | Predicting customer churn next quarter |
| **Prescriptive** | *What should we do?* | Recommends actions to achieve desired outcomes | Suggesting optimal pricing to maximize profit |

### Detailed Descriptions:

**1. Descriptive Analytics**  
Uses historical data to describe what has happened. Techniques include averages, totals, and data visualization.  
*Example:* A dashboard showing total revenue per month for the last year.

**2. Diagnostic Analytics**  
Digs deeper to understand *why* something happened by finding correlations and patterns.  
*Example:* Determining that a sales drop was caused by a competitor's promotion.

**3. Predictive Analytics**  
Uses statistical models and machine learning to forecast future events.  
*Example:* Using past purchase behavior to predict which customers are likely to buy again.

**4. Prescriptive Analytics**  
Goes beyond prediction to recommend specific actions. Often uses optimization and simulation.  
*Example:* A logistics system recommending the optimal delivery route to minimize cost and time.

## Question 3: Explain the difference between structured, semi-structured, and unstructured data. Provide examples.

### 1. Structured Data
- Highly organized data stored in a fixed format (rows and columns).
- Easily searchable using SQL.
- Stored in relational databases.
- **Examples:** Excel spreadsheets, SQL database tables, CSV files (e.g., a customer table with Name, Age, Email columns).

### 2. Semi-Structured Data
- Does not conform to a rigid tabular structure but has some organizational properties (tags, markers).
- Self-describing structure.
- **Examples:** JSON files, XML files, HTML, email (has headers like To/From but unstructured body), NoSQL databases.

```json
{
  "name": "Alice",
  "age": 30,
  "hobbies": ["reading", "cycling"]
}
```

### 3. Unstructured Data
- No predefined format or organization. Hardest to analyze.
- Requires NLP, image processing, or other AI techniques.
- **Examples:** Social media posts, images, videos, audio recordings, PDFs, Word documents.

| Feature | Structured | Semi-Structured | Unstructured |
|---------|------------|-----------------|---------------|
| Format | Fixed schema | Flexible schema | No schema |
| Searchability | Easy (SQL) | Moderate | Difficult |
| Storage | RDBMS | NoSQL, XML/JSON | Data lakes |
| % of world data | ~20% | ~15% | ~65% |

## Question 4: Describe the data analysis process (data collection, cleaning, transformation, modeling, interpretation, reporting)

The data analysis process follows a series of stages:

### Stage 1: Data Collection
Gathering relevant data from various sources.
- Sources: databases, APIs, web scraping, surveys, sensors, social media.
- Key concern: Ensure data is relevant, sufficient, and reliable.

### Stage 2: Data Cleaning
Identifying and fixing errors, inconsistencies, and missing values.
- Removing duplicates
- Handling missing values (imputation or removal)

- Removing outliers

Converting data into a suitable format for analysis.
- Normalization/standardization
- Encoding categorical variables

Applying statistical or machine learning models to extract insights.
- Regression analysis, clustering, classification
- Identifying patterns and relationships

### Stage 5: Interpretation
Making sense of the results from the model.
- Answering the original business question
- Identifying actionable insights
- Evaluating confidence in findings

### Stage 6: Reporting
Communicating findings to stakeholders.
- Creating dashboards, charts, and presentations
- Writing reports with clear recommendations
- Tailoring the message to the audience

## Question 5: What is data cleaning? Explain common data quality issues.

**Data Cleaning** (also called data cleansing or scrubbing) is the process of detecting and correcting (or removing) corrupt, inaccurate, or irrelevant records from a dataset to improve data quality.

### Common Data Quality Issues:

| Issue | Description | Example |
|-------|-------------|---------|
| **Missing Values** | Fields with no data | Age column has blank entries |
| **Duplicates** | Same record entered more than once | Customer "John Smith" appears twice |
| **Inconsistent Formatting** | Same data in different formats | Dates as "01/01/2024" vs "Jan 1, 2024" |
| **Outliers/Errors** | Values far outside expected range | Age = 999 |
| **Typos** | Misspellings | "Femal" instead of "Female" |
| **Irrelevant Data** | Data that doesn't add value to analysis | Unrelated columns |
| **Inaccurate Data** | Data that is factually wrong | Wrong phone number |
| **Structural Errors** | Wrong data types | Age stored as text instead of integer |

### Data Cleaning Techniques:
```python
import pandas as pd

df = pd.read_csv('data.csv')

# Check for missing values
print(df.isnull().sum())

# Fill missing values
df['Age'].fillna(df['Age'].mean(), inplace=True)

# Remove duplicates
df.drop_duplicates(inplace=True)

# Fix inconsistent formatting
df['Gender'] = df['Gender'].str.lower().str.strip()
```

## Question 6: Explain the concept of data visualization and its role in communicating insights.

**Data Visualization** is the graphical representation of data and information using visual elements like charts, graphs, and maps to make complex data more accessible and understandable.

### Why Data Visualization Matters:
- **Human brains process visuals 60,000x faster** than text.
- Transforms raw numbers into meaningful patterns.
- Reveals trends, outliers, and relationships that are hard to spot in tables.

### Role in Communicating Insights:

## Question 7: Differentiate between Bar chart, Line graph, Pie chart, Histogram, and Scatter plot. State when each should be used.

### 1. Bar Chart
- **What it shows:** Compares discrete categories using rectangular bars.
- **Use when:** Comparing values across different categories (e.g., sales by product, revenue by region).
- **Example:** Sales figures for 5 different products.

### 2. Line Graph
- **What it shows:** Shows trends over time using connected data points.
- **Use when:** Displaying continuous data over time (e.g., stock prices, temperature changes).
- **Example:** Monthly website traffic over a year.

### 3. Pie Chart
- **What it shows:** Shows parts-to-whole relationships as slices of a circle.
- **Use when:** Displaying proportions of a whole (ideally ≤6 categories).
- **Example:** Market share breakdown among competitors.

### 4. Histogram
- **What it shows:** Displays the distribution/frequency of a continuous variable using bars (no gaps).
- **Use when:** Understanding the distribution or spread of numerical data.
- **Example:** Distribution of student exam scores.

### 5. Scatter Plot
- **What it shows:** Shows the relationship/correlation between two continuous variables.
- **Use when:** Exploring correlation or identifying clusters/outliers between two variables.
- **Example:** Relationship between advertising spend and revenue.

| Chart Type | Best For | X-Axis | Y-Axis |
|------------|----------|--------|--------|
| Bar Chart | Category comparison | Categories | Values |
| Line Graph | Trends over time | Time | Values |
| Pie Chart | Part-to-whole | N/A | Proportions |
| Histogram | Data distribution | Bins (ranges) | Frequency |
| Scatter Plot | Correlation | Variable 1 | Variable 2 |

```python
import matplotlib.pyplot as plt
import numpy as np

fig, axes = plt.subplots(2, 3, figsize=(15, 8))

# Bar Chart
products = ['A', 'B', 'C', 'D']
sales = [400, 300, 500, 200]
axes[0,0].bar(products, sales, color='steelblue')
axes[0,0].set_title('Bar Chart - Product Sales')

# Line Graph
months = range(1, 13)
revenue = [100, 120, 115, 130, 145, 160, 155, 170, 180, 200, 195, 210]
axes[0,1].plot(months, revenue, marker='o', color='green')
axes[0,1].set_title('Line Graph - Monthly Revenue')

# Pie Chart
shares = [40, 30, 20, 10]
labels = ['Company A', 'Company B', 'Company C', 'Others']
axes[0,2].pie(shares, labels=labels, autopct='%1.1f%%')
axes[0,2].set_title('Pie Chart - Market Share')

# Histogram
scores = np.random.normal(70, 15, 200)
axes[1,0].hist(scores, bins=15, color='orange', edgecolor='black')
axes[1,0].set_title('Histogram - Exam Score Distribution')

# Scatter Plot
ad_spend = np.random.randint(1000, 10000, 50)
revenue2 = ad_spend * 2.5 + np.random.normal(0, 2000, 50)
axes[1,1].scatter(ad_spend, revenue2, color='red', alpha=0.6)
axes[1,1].set_title('Scatter Plot - Ad Spend vs Revenue')

axes[1,2].axis('off')
plt.tight_layout()
plt.show()
```

## Question 8: Define the following statistical terms: Mean, Median, Mode, Standard Deviation, Variance

### 1. Mean (Arithmetic Average)
The sum of all values divided by the number of values.

$$\bar{x} = \frac{\sum x_i}{n}$$

**Example:** Data = [2, 4, 6, 8, 10] → Mean = (2+4+6+8+10)/5 = **6**  
**Best for:** Symmetric distributions without extreme outliers.

---
### 2. Median
The middle value when data is sorted in order. If even number of values, it's the average of the two middle values.

**Example:** Data = [1, 3, 5, 7, 9] → Median = **5**  
Data = [1, 3, 5, 7] → Median = (3+5)/2 = **4**  
**Best for:** Skewed distributions or when outliers are present (e.g., income data).

---
### 3. Mode
The value that appears most frequently in a dataset. A dataset can have no mode, one mode, or multiple modes.

**Example:** Data = [1, 2, 2, 3, 4, 4, 4] → Mode = **4**  
**Best for:** Categorical data or finding the most common value.

---
### 4. Standard Deviation
Measures how spread out values are around the mean. Low SD = data close to mean; High SD = data spread out.

$$\sigma = \sqrt{\frac{\sum(x_i - \bar{x})^2}{n}}$$

**Example:** Scores [70, 72, 68, 71, 69] have a low SD; scores [50, 90, 60, 80, 70] have a high SD.  
**Best for:** Understanding variability and consistency.

---
### 5. Variance
The square of the standard deviation. Represents the average squared deviation from the mean.

$$\sigma^2 = \frac{\sum(x_i - \bar{x})^2}{n}$$

**Note:** Variance is in squared units; SD converts it back to original units.

````python
import numpy as np
from scipy import stats

data = [4, 8, 6, 5, 3, 2, 8, 9, 2, 5]

print(f"Data: {data}")
print(f"Mean:               {np.mean(data):.2f}")
print(f"Median:             {np.median(data):.2f}")
print(f"Mode:               {stats.mode(data, keepdims=True).mode[0]}")
print(f"Standard Deviation: {np.std(data):.2f}")
print(f"Variance:           {np.var(data):.2f}")
````

## Question 9: Explain correlation vs. causation with examples.

### Correlation
**Correlation** is a statistical relationship between two variables — when one changes, the other tends to change too. It does NOT imply one causes the other.

- Measured by the **correlation coefficient (r)**: ranges from -1 to +1.
  - r = +1: Perfect positive correlation
  - r = -1: Perfect negative correlation  
  - r = 0: No correlation

### Causation
**Causation** (or causality) means that one variable *directly causes* a change in another variable.

### Key Principle: "Correlation does not imply causation"

| | Correlation | Causation |
|--|-------------|----------|
| **Definition** | Variables move together | One variable causes the other |
| **Proven by** | Statistical analysis | Controlled experiments |
| **Direction** | Can be bidirectional | Has a clear direction |

### Examples:

**Example 1 — Correlation WITHOUT Causation (Spurious):**  
Ice cream sales and drowning rates are positively correlated. But ice cream doesn't cause drowning — both are caused by a **third variable (confound)**: hot weather.

**Example 2 — Correlation WITH Causation:**  
Smoking is correlated with lung cancer. Extensive research (controlled experiments and biological mechanisms) has proven that smoking *causes* lung cancer.

**Example 3 — Business Context:**  
A company notices that website visits and sales are correlated. But does more traffic *cause* more sales, or are both driven by a marketing campaign? Further investigation (A/B testing) is needed to establish causation.

### How to Establish Causation:
1. Randomized Controlled Trials (RCTs)
2. A/B Testing
3. Natural experiments
4. Regression discontinuity design

## Question 10: What is a dashboard? List key characteristics of an effective business dashboard.

### What is a Dashboard?
A **dashboard** is a visual display of the most important information needed to achieve one or more objectives, consolidated on a single screen so the information can be monitored at a glance.

It is a real-time, interactive interface that aggregates and presents data from multiple sources using charts, graphs, metrics, and tables.

**Types of Dashboards:**
- **Operational:** Real-time monitoring (e.g., server uptime, call center metrics)
- **Analytical:** Deep-dive analysis (e.g., marketing performance)
- **Strategic/Executive:** High-level KPIs for leadership (e.g., company revenue, customer growth)

### Key Characteristics of an Effective Business Dashboard:

1. **Clarity & Simplicity:** Uncluttered design; information is easy to understand at a glance.
2. **Relevance:** Only shows data relevant to the audience's goals and decisions.
3. **Real-time / Up-to-date Data:** Refreshes automatically to reflect the latest information.
4. **Interactivity:** Allows users to drill down, filter, and explore data dynamically.
5. **Appropriate Visualizations:** Uses the right chart types for each metric.
6. **Contextual Information:** Shows benchmarks, targets, or comparisons (e.g., vs. last year).
7. **Consistent Design:** Uses a coherent color scheme, font, and layout.
8. **Mobile-Friendly:** Accessible on different devices.
9. **Actionable Insights:** Enables users to make decisions directly from the information shown.
10. **Performance:** Loads quickly even with large datasets.

**Popular Dashboard Tools:** Tableau, Power BI, Google Data Studio (Looker), Metabase, Grafana.

## Question 11: Discuss the importance of data ethics and privacy in data analysis.

### Data Ethics
**Data ethics** refers to the moral obligations that govern how data is collected, stored, processed, shared, and used. It involves ensuring data practices are fair, transparent, and respectful of individuals' rights.

### Data Privacy
**Data privacy** is the right of individuals to control how their personal information is collected and used.

### Why They Matter:

#### 1. Trust and Reputation
Organizations that handle data ethically build trust with customers, employees, and partners. A single breach can permanently damage reputation (e.g., Facebook-Cambridge Analytica scandal).

#### 2. Legal Compliance
Regulations require responsible data handling:
- **GDPR** (EU): Fines up to €20 million or 4% of global revenue.
- **POPIA** (South Africa): Protects personal information of South Africans.
- **CCPA** (California): Consumer privacy rights.
- **HIPAA** (USA): Healthcare data protection.

#### 3. Preventing Harm
Unethical data use can lead to discrimination (e.g., biased algorithms in hiring/lending), identity theft, and manipulation.

#### 4. Fairness and Bias
Data analysts must ensure models don't perpetuate biases against protected groups (race, gender, age).

### Key Ethical Principles in Data Analysis:
- **Transparency:** Be open about how data is collected and used.
- **Consent:** Obtain explicit permission before collecting personal data.
- **Data Minimization:** Collect only what's necessary.
- **Anonymization:** Remove identifying information when possible.
- **Security:** Protect data from breaches and unauthorized access.
- **Accountability:** Take responsibility for data decisions and their consequences.

## Question 12: Explain the concept of data normalization and why it is important.

### What is Data Normalization?
**Data normalization** is the process of rescaling numeric features so they are on a similar scale, typically between 0 and 1 or with a mean of 0 and standard deviation of 1.

There are two main types:

#### 1. Min-Max Normalization (Scaling)
Scales data to a fixed range, usually [0, 1].

$$x_{norm} = \frac{x - x_{min}}{x_{max} - x_{min}}$$

#### 2. Z-Score Standardization
Transforms data to have mean = 0 and standard deviation = 1.

$$z = \frac{x - \mu}{\sigma}$$

### Why is Normalization Important?

1. **Prevents Feature Dominance:** Without normalization, features with large scales (e.g., income: 50,000) dominate features with small scales (e.g., age: 30) in algorithms.
2. **Improves Model Performance:** Machine learning algorithms like K-Nearest Neighbors, SVM, and Neural Networks are sensitive to feature scales.
3. **Faster Convergence:** Gradient descent converges faster with normalized data.
4. **Fair Comparison:** Allows meaningful comparison between variables with different units.

**Example Without Normalization:**  
Salary (range: 20,000–200,000) would completely overshadow Age (range: 18–65) in a distance-based algorithm.

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler
import numpy as np

data = np.array([[1000, 25], [5000, 30], [2500, 22], [8000, 45]])

# Min-Max Normalization
scaler = MinMaxScaler()
normalized = scaler.fit_transform(data)
print("Min-Max Normalized:")
print(normalized)

# Z-Score Standardization
std_scaler = StandardScaler()
standardized = std_scaler.fit_transform(data)
print("\nStandardized (Z-Score):")
print(standardized)
```

## Question 13: Describe the difference between qualitative and quantitative data.

### Qualitative Data (Categorical)
Describes qualities or characteristics. It is **non-numerical** and represents categories or labels.

**Sub-types:**
- **Nominal:** Categories with no natural order (e.g., colors, gender, country).
- **Ordinal:** Categories with a meaningful order (e.g., satisfaction ratings: poor/average/good/excellent).

**Examples:**
- Customer feedback (positive/negative/neutral)
- Product color (red, blue, green)
- Education level (high school, bachelor's, master's)

---

### Quantitative Data (Numerical)
Represents measurable quantities. It is **numerical** and can be subjected to arithmetic operations.

**Sub-types:**
- **Discrete:** Countable, whole numbers (e.g., number of customers, goals scored).
- **Continuous:** Can take any value within a range (e.g., temperature, height, weight).

**Examples:**
- Sales revenue ($45,230)
- Number of employees (150)
- Customer age (28.5 years)

---

| Feature | Qualitative | Quantitative |
|---------|-------------|---------------|
| Nature | Descriptive, categorical | Numerical, measurable |
| Examples | Gender, color, occupation | Age, salary, temperature |
| Analysis | Frequency counts, mode | Mean, median, SD, regression |
| Visualization | Bar charts, pie charts | Histograms, scatter plots, line graphs |
| Sub-types | Nominal, Ordinal | Discrete, Continuous |

## Question 14: What are Key Performance Indicators (KPIs)? Give examples in a business context.

### What are KPIs?
**Key Performance Indicators (KPIs)** are measurable values that demonstrate how effectively a company or individual is achieving key business objectives. They help organizations track progress, evaluate success, and make data-driven decisions.

### Characteristics of Good KPIs (SMART):
- **S**pecific – Clearly defined
- **M**easurable – Quantifiable
- **A**chievable – Realistic
- **R**elevant – Aligned with business goals
- **T**ime-bound – Has a timeframe

### KPI Examples by Business Department:

| Department | KPI | Description |
|------------|-----|-------------|
| **Sales** | Monthly Revenue | Total revenue generated per month |
| **Sales** | Conversion Rate | % of leads that become customers |
| **Marketing** | Customer Acquisition Cost (CAC) | Cost to acquire one new customer |
| **Marketing** | Return on Ad Spend (ROAS) | Revenue per $1 spent on advertising |
| **Customer Service** | Customer Satisfaction Score (CSAT) | Average satisfaction rating |
| **Customer Service** | First Response Time | Avg time to respond to a customer query |
| **Finance** | Gross Profit Margin | (Revenue - COGS) / Revenue × 100 |
| **Operations** | Employee Productivity | Output per employee per period |
| **HR** | Employee Turnover Rate | % of employees who leave per year |
| **E-commerce** | Cart Abandonment Rate | % of users who add to cart but don't buy |

### Example:
> **KPI:** Monthly Recurring Revenue (MRR)  
> **Target:** R500,000/month  
> **Current:** R420,000/month  
> **Status:** 84% of target — action needed to close the gap.

## Question 15: Explain the concept of data storytelling.

### What is Data Storytelling?
**Data storytelling** is the practice of building a compelling narrative around data analysis to communicate insights effectively to a target audience. It combines three elements:

1. **Data** — The factual foundation (statistics, analysis results)
2. **Visuals** — Charts, graphs, and images that make the data digestible
3. **Narrative** — The story that contextualizes the data and guides the audience to a conclusion

### Why Data Storytelling Matters:
- Raw data alone doesn't drive action — stories do.
- Decision-makers often lack deep technical knowledge; a story bridges the gap.
- Emotionally engaging narratives make insights more memorable.
- Helps stakeholders understand *what the data means* and *what to do about it*.

### Elements of Effective Data Storytelling:

| Element | Description |
|---------|-------------|
| **Know Your Audience** | Tailor technical level and focus to the viewers |
| **Start with a Hook** | Begin with the most compelling insight or a question |
| **Provide Context** | Show benchmarks, comparisons, or historical trends |
| **Use the Right Visuals** | Match chart types to the message |
| **Highlight Key Insights** | Guide attention to what matters most |
| **End with a Call to Action** | Clearly state recommendations or next steps |

### Example of Data Storytelling:
> **Without storytelling:** "Customer churn increased from 5% to 12% between Q1 and Q3."
>
> **With storytelling:** "We're losing customers at an alarming rate — churn has more than *doubled* in just 6 months. Data shows that customers who don't engage with our app in the first 7 days after signup are 3x more likely to leave. If we implement an onboarding email campaign targeting these users, we project recovering R2 million in annual revenue."

### The Data Storytelling Framework:
```
📊 DATA → 🔍 INSIGHT → 📖 NARRATIVE → 💡 ACTION
```

**Tools for Data Storytelling:** Tableau, Power BI, Google Slides, Canva, Jupyter Notebooks (with visualizations).
